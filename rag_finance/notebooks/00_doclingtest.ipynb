```python
from docling.document_converter import DocumentConverter
```

```python
converter = DocumentConverter()
```

```python
result = converter.convert(
    "../data/raw/3m_2018.pdf"
)
```

```python
type(result.document)
```

```python
from pathlib import Path

save_dir = Path("../data/parsed")
save_dir.mkdir(parents=True, exist_ok=True)

result.document.save_as_json(
    save_dir / "3m_2018_docling.json"
)

print("Saved!")
```

```python
doc = result.document
```

```python
text_blocks = []
table_blocks = []

for text in doc.texts:
    text_blocks.append(text)

for table in doc.tables:
    table_blocks.append(table)
```

```python
import pickle

with open("../data/parsed/text_blocks.pkl", "wb") as f:
    pickle.dump(text_blocks, f)

with open("../data/parsed/table_blocks.pkl", "wb") as f:
    pickle.dump(table_blocks, f)
```
